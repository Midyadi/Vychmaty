## **$$\textbf{Решение}$$**

По условию в уравнении $\frac{d \mathbf{y}}{d t}=\boldsymbol{\Lambda}_{x x} \mathbf{y}+\mathbf{f}(t)$ матрица  $\boldsymbol{\Lambda}_{x x}$ **трёхдиагональна**, а это значит, что для приближённого вычисления второй проихводной используется только соседние точки. Это соответствует схеме $$\left(\frac{\partial^2 y}{\partial x^2}\right)_n = \dfrac{y(x_{n+1})-2y(x_n)+y(x_{n-1})}{h^2}+\mathcal{O}(h^2)$$ имеющей второй порядок аппроксимации.

Неявный метод Рунге-Кутты с тремя стадиями (который дан по условию варианта) имеет порядок аппроксимации $\geq$ числу стадий, то есть $\geq 3$. Это значит, что основная ошибка будет идти не от метода Рунге-Кутты, а от аппроксимации производной, то есть на выходе ожидаем порядок сходимости $p=2$ по координате.

Возьмём простейшую аппроксимацию производной по времени $$\left(\frac{\partial y}{\partial t}\right)_n = \dfrac{y^{(t+\tau)}(x_{n})-y^{(t)}(x_{n})}{\tau}$$

Также заметим, что явному методу Рунге-Кутты необходимо условие $\tau \propto h^2$ (шаг сетки по времени пропорционален квадрату шага по координате) для устойчивости (так как в формуле происходит умножение на коэффициент $\sigma = \dfrac{\tau}{h^2}$), в то время как в нашем случае (неявный метод) такого условия не требуется (так как в неявной схеме умножение идёт не на $\sigma$, а на $1\pm\sigma$), поэтому при изменении координатной сетки можно не трогать временную сетку.

Но для оценки порядка сходимости стоит сохранять пропорцию $\dfrac{\tau}{h^2}$, чтобы создать одинаковые условия для разных шагов.

Также нужно определить граничные условия (так как не можем посчитать бесконечное число точек). Можно воспользоваться известным точным решением, однако лучше считать точное решение на данном этапе неизвестным и самим выбрать начальные условия.

Например, можно "отойти" достаточно далеко от нуля и задать нулевые граничные условия, которые в силу изначального распределения $\varphi(x) = e^{-x^2}$ будут почти идентичны реальным значениям функции в этой точке на протяжении всего отрезка времени. Кроме того, мы не нарушим порядок сходимости.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.sparse import diags, identity
from scipy.sparse.linalg import spsolve
from matplotlib import cm


L = 10 # границы пространственной области [-10,10], так как e^(-10^2) = 3.7*10^(-44) - почти ноль
T =  1 # Границы временной области [0,10] 
h = 0.05    # Пространственный шаг
tau = h**2/4    # Временной шаг, пропорциональный h^2
xs = np.linspace(-L, L, int(2*L/h)+1)
ts = np.linspace(0,T, int(T/tau)+1)

# Начальное условие
phi = np.exp(-xs**2)

# Точное решение
def exact_solution(x, t):
    return np.exp(-x**2 / (4 * t + 1)) / np.sqrt(4 * t + 1)

# Таблица Бутчера
butch_table = np.array([[1/3, 1/3, 0,0],[1,1/3,2/3,0],[1,0,0,1],[0,3/4,3/4,-1/2]])

A = diags([1, -2, 1], [-1, 0, 1], shape=(len(xs), len(xs)), format="csr") / h**2
I = identity(len(xs), format='csr')

def f(u):
    '''Считает аппроксимацию второй производной по координате
    '''
    centr_diff = (A @ u)[1:-1]
    left_diff = (2*u[0]-5*u[1]+4*u[2]-u[3])/h**2
    right_diff = (2*u[-1]-5*u[-2]+4*u[-3]-u[-4])/h**2
    return np.concatenate(([left_diff], centr_diff, [right_diff]), axis=None)


plt.figure(figsize=(12, 8))

u = np.zeros((len(ts), len(xs)))
u[0, :] = phi


for i,t in enumerate(ts[:-1]):
    u_cur = u[i, :]
    
    # Считаем коэффициенты k_i
    b1 = f(u_cur)
    k1 = spsolve(I - tau * butch_table[0,1] * A, b1)
    
    b2 = f(u_cur + tau * butch_table[1,1] * k1)
    k2 = spsolve(I - tau * butch_table[1,2] * A, b2)

    b3 = f(u_cur + tau * (butch_table[2,1] * k1 + butch_table[2,2] * k2))
    k3 = spsolve(I - tau * butch_table[2,3] * A, b3) 
    
    # Заполняем таблицу значений
    u[i+1, :] = u_cur + tau * (butch_table[3,1] * k1 + butch_table[3,2] * k2 + butch_table[3,3] * k3)
    
    u[i+1, 0] = 0
    u[i+1, -1] = 0
    
    # Отрисовываем некоторые моменты времени
    if i % 150 == 0 or i == len(ts)-2:
        color = cm.jet(i/(len(ts)-2))
        color2 = cm.jet(1-i/(len(ts)-2))
        plt.plot(xs, u[i, :], label=f'Численное решение (T={round(t,3)})', color = color)
        plt.plot(xs, exact_solution(xs, t), '--', label=f'Точное решение (T={round(t,3)})', color = "crimson")

plt.xlabel('x')
plt.ylabel('u(x, t)')
plt.legend()
plt.grid()
plt.show()
```

Теперь проверим порядок сходимости метода по частоте разбиения координатной сетки (для простоты проделаем это только для одного момента времени, например, $t = 1$ с)

```python
# Проверка порядка сходимости (для T=1)
T = 1
hs = [1, 0.5,0.25,0.125,0.0625, 0.03125, 0.015625, 0.078125]
errors = []

for h in hs:
    tau = h**2/4
    xs = np.linspace(-L, L, int(2*L/h)+1)
    ts = np.linspace(0,T, int(T/tau)+1)

    # Начальное условие
    phi = np.exp(-xs**2)
        
    A = diags([1, -2, 1], [-1, 0, 1], shape=(len(xs), len(xs)), format="csr") / h**2
    I = identity(len(xs), format='csr')
    
    u = np.zeros((len(ts), len(xs)))
    u[0, :] = phi
    
    for i,t in enumerate(ts[:-1]):
        u_cur = u[i, :]
        
        # Считаем коэффициенты k_i
        b1 = f(u_cur)
        k1 = spsolve(I - tau * butch_table[0,1] * A, b1)
        
        b2 = f(u_cur + tau * butch_table[1,1] * k1)
        k2 = spsolve(I - tau * butch_table[1,2] * A, b2)

        b3 = f(u_cur + tau * (butch_table[2,1] * k1 + butch_table[2,2] * k2))
        k3 = spsolve(I - tau * butch_table[2,3] * A, b3) 
        
        # Заполняем таблицу значений
        u[i+1, :] = u_cur + tau * (butch_table[3,1] * k1 + butch_table[3,2] * k2 + butch_table[3,3] * k3)
        
        u[i+1, 0] = 0
        u[i+1, -1] = 0
    
    u_exact = exact_solution(xs, T)

    # Возьмём максимальную норму max(abs(u-u_exact))
    errors.append(np.linalg.norm(u[-1] - u_exact, ord=np.inf))

# 
print("\nОпределение порядка сходимости:")
for i, error in enumerate(errors[:-1]):
    p = np.log2(error/errors[i+1])
    print(f"Изменение шага координатной сетки h: {hs[i]}->{hs[i+1]};  p = {round(p,4)}")
```

При уменьшении шага координатной сетки действительно приближаемся к предсказанному порядку сходимости $p=2$. При последнем уменьшении шага, вероятно, была достигнута критическая величина шага (причём вероятнее всего по времени, так как $\tau \propto h^2$) и произошло накопление машинной погрешности, что исказило порядок сходимости.

## Задача 2. Уравнение акустики (варианты 1-12).


Проверить, корректно ли поставлена краевая задача для системы уравнений акустики

$$
u_t+v_x=0, v_t+u_x=0
$$


при начальных условиях $u(x, 0)=f(x), v(x, 0)=g(x)$ и краевых условиях согласно варианту. При необходимости исправить ошибки и решить систему, предложив подходящую разностную схему. Для оценки порядка сходимости схемы дважды использовать сгущение сеток в два раза. Оценить реальную скорость сходимости. Сравнить численное решение с точным.

## **$$\textbf{Решение}$$**

Для начала решим уравнение точно:

В ходе решения помимо собственно ответов получили ещё и несколько условий на функции $f(x)$ и $g(x)$ из начальных условий: $$f(x) - g(x) = const$$ $$f'(0) = g'(0) = 0$$

В качестве разностной схемы возьмём самую простую (явный левый уголок). Это значит, что после аппроксимаций производных наша система уравнений будет выглядеть следующим образом: $$\dfrac{u^{j+1}_i - u^{j}_i}{\tau}+ \dfrac{v^{j}_i - v^{j}_{i-1}}{h}+\mathcal{O}(\tau + h)=0$$ $$\dfrac{v^{j+1}_i - v^{j}_i}{\tau}+ \dfrac{u^{j}_i - u^{j}_{i-1}}{h}+\mathcal{O}(\tau + h)=0$$ Здесь верхний индекс - временной, а нижний - пространственный. Тогда ожидаем порядок сходимости $p = 1$.

```python
# Раз уж функции f и g не хаданы по условию,
# а явно посчитать что-то хочется, то зададим их сами:
def f(x): 
    return np.exp(-(x/5)**2)*np.cos(5*x)**2

def g(x): 
    return np.exp(-(x/5)**2)*np.cos(5*x)**2


# Задали f и g таким образом, чтобы, во-первых, выполнялись условия разрешимости 
# системы уравнений, а во-вторых, чтобы на бесконечности они обе стемились к нулю,
# чтобы можно было как в прошлой задаче без особого ущерба задать нулевые граничные условия
# при переходе к конечному отребку по координате. 

# Параметры задачи
L = 10
T = 2
h = 0.01
tau = h/2  # для выбранной схемы устойчивость достигается при tau<=h

# Сетки
xs = np.linspace(0, L, int(L/h)+1)
ts = np.linspace(0, T, int(T/tau)+1)
u = np.zeros((len(ts), len(xs)))
v = np.zeros((len(ts), len(xs)))

# Точные решения:
def u_exa(coords,times):
    u_ex = np.zeros((len(times), len(coords)))
    for i, x in enumerate(coords):
        for j,t in enumerate(times):
            if x>t:
                u_ex[j, i] = f(x-t)
            else:
                u_ex[j, i] = f(0)
    return u_ex

def v_exa(coords,times):
    v_ex = np.zeros((len(times), len(coords)))
    for i, x in enumerate(coords):
        for j,t in enumerate(times):
            if x>t:
                v_ex[j, i] = g(x-t)
            else:
                v_ex[j, i] = g(0)
    return v_ex

# Начальные условия
u[0,:] = f(xs)
v[0,:] = g(xs)

for i, t in enumerate(ts[1:]):

    # Граничные условия
    u[i+1,0] = f(0)
    v[i+1,0] = g(0)
    
    # Внутренние точки
    for j, x in enumerate(xs[1:-1]):
        u[i+1,j+1] = u[i,j+1] - tau/h*(v[i,j+1] - v[i,j])
        v[i+1,j+1] = v[i,j+1] - tau/h*(u[i,j+1] - u[i,j])
    
    # Нулевые граничные условия справа (ограничевающие область расчётов)
    u[i+1,-1] = 0
    v[i+1,-1] = 0

# Визуализация
plt.figure(figsize=(12,6))
plt.subplot(121)
plt.ylabel('t')
plt.xlabel('x')
plt.imshow(u, extent=[0,L,0,T], aspect='auto', origin='lower')
plt.colorbar(label='u(x,t)')
plt.title('График u(x,t)')

plt.subplot(122)
plt.ylabel('t')
plt.xlabel('x')
plt.imshow(v, extent=[0,L,0,T], aspect='auto', origin='lower')
plt.colorbar(label='v(x,t)')
plt.title('График v(x,t)')
plt.show()
```

Нарисуем график точного решения

```python
u_ex = u_exa(xs, ts)
v_ex = v_exa(xs, ts)

plt.figure(figsize=(12,6))
plt.subplot(121)
plt.ylabel('t')
plt.xlabel('x')
plt.imshow(u_ex, extent=[0,L,0,T], aspect='auto', origin='lower')
plt.colorbar(label='u(x,t)')
plt.title('График точного решения u(x,t)')

plt.subplot(122)
plt.ylabel('t')
plt.xlabel('x')
plt.imshow(v_ex, extent=[0,L,0,T], aspect='auto', origin='lower')
plt.colorbar(label='v(x,t)')
plt.title('График точного решения v(x,t)')
plt.show()
```

Видим, что визуально графики точного и приближённого решений совпали.

Оценим порядок сходимости:

```python
# Проверка порядка сходимости для T=1
T = 1
hs = [0.025, 0.0125, 0.00625, 0.003125,0.0015625]
u_errors = []
v_errors = []

for h in hs:
    # Параметры задачи
    L = 10
    T = 2
    tau=h/2

    # Сетки
    xs = np.linspace(0, L, int(L/h)+1)
    ts = np.linspace(0, T, int(T/tau)+1)
    u = np.zeros((len(ts), len(xs)))
    v = np.zeros((len(ts), len(xs)))

    # Начальные условия
    u[0,:] = f(xs)
    v[0,:] = g(xs)

    for i, t in enumerate(ts[1:]):

        # Граничные условия
        u[i+1,0] = f(0)
        v[i+1,0] = g(0)
        
        # Внутренние точки
        for j, x in enumerate(xs[1:-1]):
            u[i+1,j+1] = u[i,j+1] - tau/h*(v[i,j+1] - v[i,j])
            v[i+1,j+1] = v[i,j+1] - tau/h*(u[i,j+1] - u[i,j])
        
        # Нулевые граничные условия справа (ограничевающие область расчётов)
        u[i+1,-1] = 0
        v[i+1,-1] = 0
    
    u_ex = u_exa(xs, ts)
    v_ex = v_exa(xs, ts)

    # Возьмём максимальную норму max(abs(u-u_exact))
    u_errors.append(np.linalg.norm(u[-1] - u_ex[-1], ord=np.inf))
    v_errors.append(np.linalg.norm(v[-1] - v_ex[-1], ord=np.inf))

print("\nОпределение порядка сходимости для функции U(x,t)")
for i in range(len(hs)-1):
    p_u = np.log2(u_errors[i]/u_errors[i+1])
    print(f"Изменение шага координатной сетки h: {hs[i]}->{hs[i+1]};  p_u = {round(p_u,4)}")

print("\nОпределение порядка сходимости для функции V(x,t)")
for i in range(len(hs)-1):
    p_v = np.log2(v_errors[i]/v_errors[i+1])
    print(f"Изменение шага координатной сетки h: {hs[i]}->{hs[i+1]};  p_v = {round(p_v,4)}")
```

Действительно, при уменьшении шага сетки в 2 раза мы приближаемся к теоретическому значению $p = 1$ для обеих функций.

Приближение к теор. значению не такое быстрое, как в прошлом номере (там получали почти ровно $p = 2$), это может быть связано с особенностями разностной схемы (например, в этом случае она явная, в то время как в прошлом номере использовали неявную схему).

## Задача 2. Смешанное уравнение гиперболического типа (варианты 13-18).

 Найти решение смешанной задачи для уравнения гиперболического типа на сетке $\left(x_m, t^n\right)$


 Использовать метод второго порядка аппроксимации. Начальную сетку выбрать самостоятельно. Для оценки порядка сходимости схемы дважды использовать сгущение сеток в два раза. Оценить реальную скорость сходимости.

Найти точное решение каждой задачи. Сравнить численное решение с точным.

**Вариант 13**

$u_{t t}^{\prime \prime}(x, t)=u_{x x}^{\prime \prime}(x, t)+\frac{2\left(a^2-1\right)}{(x+a t+1)^3}, \quad 0 \leq x \leq 1, \quad 0 \leq t \leq 1$,

$$
\begin{aligned}
& u(0, t)=\frac{1}{a t+1}, \quad u(1, t)=\frac{1}{a t+2}, \quad 0 \leq t \leq 1 \\
& u(x, 0)=\frac{1}{x+1}, \quad u_t^{\prime}(x, 0)=-\frac{a}{(x+1)^2}, \quad 0 \leq x \leq 1 \\
& a=0.5+0.1 k(k=0,1, \ldots, 10)
\end{aligned}
$$

**Вариант 14**

\end{aligned}
$$

**Вариант 15**

$u_{t t}^{\prime \prime}(x, t)=u_{x x}^{\prime \prime}(x, t)+\frac{2 a}{(a(x+t)+2)^2}, \quad 0 \leq x \leq 1, \quad 0 \leq t \leq 1$,

$$
\begin{aligned}
& u(0, t)=0, \quad u(1, t)=\frac{1}{a(t+1)+2}, \quad 0 \leq t \leq 1 \\

\end{aligned}
$$
